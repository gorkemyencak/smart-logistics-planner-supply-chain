```python
import json
from pathlib import Path
import sys
import os

PROJECT_ROOT = os.path.abspath("..")
sys.path.append(PROJECT_ROOT)

from src.visualizer.global_visualizer import GlobalVisualizer
```

```python
# helper loader
def load_routes(cluster_id):
    path = Path(f"../data/outputs/routes/cluster_{cluster_id}_routes.json")

    if not path.exists():
        print(f"Missing {path}")
        return []
    
    with open(path) as f:
        return json.load(f)
```

```python
# loading all clusters
routes_path = Path(f"../data/outputs/routes")
n_clusters = sum(1 for f in routes_path.iterdir() if f.is_file())

all_cluster_routes = []

for c in range(n_clusters):
    routes = load_routes(c)
    if routes:
        all_cluster_routes.append(routes)

print("Loaded clusters:", len(all_cluster_routes))

# building global map
m = GlobalVisualizer.build(all_cluster_routes, save_path = '../data/outputs/map/global_map.html')
```

```python
import sys
import os
import pandas as pd
PROJECT_ROOT = os.path.abspath("..")
sys.path.append(PROJECT_ROOT)

from src.visualizer.route_visualizer import RouteVisualizer
from src.optimization.vrp_solver import VRPsolver
from IPython.display import IFrame, display, HTML
from src.utils.route_io import load_routes
```

```python
df = pd.read_csv("../data/final/smart_logistics_dataset_with_clusters.csv")
df.head()
```

```python
# sample cluster
cluster_id = 0
cluster_df = df[df['Cluster_ID'] == cluster_id].copy().reset_index(drop=True)
cluster_df
```

```python
# Load routes
routes = load_routes(cluster_id = cluster_id)
```

```python
# artificial depot row
depot_lat = cluster_df['Latitude'].mean()
depot_lon = cluster_df['Longitude'].mean()

depot_row = {
    'Latitude': depot_lat,
    'Longitude': depot_lon,
    'Demand_Forecast': 0,
    'Traffic_Status': cluster_df['Traffic_Status'].iloc[0],
    'Waiting_Time': cluster_df['Waiting_Time'].median(),
    'Traffic_Encoded': cluster_df['Traffic_Encoded'].mean(),
    'Cluster_ID': cluster_df['Cluster_ID'].mean()
}

cluster_df_with_depot = pd.concat(
    [cluster_df, pd.DataFrame([depot_row])],
    ignore_index = True
)

cluster_df_with_depot
```

```python
# Visualizer
visualizer = RouteVisualizer(depot_lat, depot_lon)

m, map_path = visualizer.visualize_cluster_routes(
    cluster_df = cluster_df, # cluster_df_with_depot,
    routes = routes,
    cluster_id = cluster_id
)

m
```

```python
# saving static image for a preview on github
visualizer.save_static_image(map_path)
```
